```python
x = "i am sushmita malakar i am data scientist i am good at painting too i am sushmita"
```

```python
from nltk.tokenize import word_tokenize
```

```python
w = word_tokenize(x)
w
```

```python
from nltk.collocations import BigramCollocationFinder, TrigramCollocationFinder, ngrams
```

```python
b = BigramCollocationFinder.from_words(w)
```

```python
b.ngram_fd
```

```python
b.ngram_fd.keys()
```

```python
t = TrigramCollocationFinder.from_words(w)
```

```python
t.ngram_fd
```

```python
n = ngrams(w, 1)
```

```python
for i in n:
    print(i)
```
