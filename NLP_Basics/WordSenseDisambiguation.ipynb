```python
x = "Sunrise (or sunup) is the moment when the upper rim of the Sun appears on the horizon in the morning, at the start of the Sun path."
```

```python
from  nltk.wsd import lesk
from nltk.tokenize import word_tokenize
```

```python
l = lesk(word_tokenize(x), "sun")
```

```python
l
```

```python
l.definition()
```
